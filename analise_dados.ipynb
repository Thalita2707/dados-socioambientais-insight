# Análise 5 - Origem das Famílias 🗺️🏡  
### Exibe as cidades/estados de origem dos entrevistados em um gráfico de barras.  
### ⚠️ Nota: Não é possível gerar um mapa real sem dados geográficos (latitude/longitude).


```python
# --- ETAPA 1: Criar uma função para limpar e padronizar as cidades de origem ---
def limpar_origem(texto):
    if not isinstance(texto, str):
        return 'Não informado'

    # Converte para minúsculas para facilitar a busca de palavras-chave
    texto_lower = texto.lower()

    # Regras de padronização para as cidades mais comuns
    if 'jardim' in texto_lower: return 'Jardim, MS'
    if 'guia lopes' in texto_lower or 'laguna' in texto_lower: return 'Guia Lopes da Laguna, MS'
    if 'bela vista' in texto_lower: return 'Bela Vista, MS'
    if 'bonito' in texto_lower: return 'Bonito, MS'
    if 'campo grande' in texto_lower: return 'Campo Grande, MS'
    if 'aquidauana' in texto_lower: return 'Aquidauana, MS'
    if 'porto murtinho' in texto_lower: return 'Porto Murtinho, MS'
    if 'gaúcha' in texto_lower or 'rio grande do sul' in texto_lower: return 'Rio Grande do Sul (Estado)'
    if 'paraguai' in texto_lower: return 'Paraguai'

    # Se não se encaixar em nenhuma regra acima, retorna o texto original de forma capitalizada
    return texto.strip().title()

# --- ETAPA 2: Aplicar a função para criar a coluna de origens padronizadas ---
df_limpo['origem_agrupada'] = df_limpo['origem_familia'].apply(limpar_origem)


# --- ETAPA 3: Agrupar as categorias menos frequentes em "Outros" ---
# Define quantas categorias principais queremos mostrar
top_n = 15
# Encontra as categorias mais comuns
top_categorias = df_limpo['origem_agrupada'].value_counts().nlargest(top_n).index

# Cria uma nova coluna para o gráfico: se a categoria não estiver no top N, ela vira "Outros"
df_limpo['origem_plot'] = df_limpo['origem_agrupada'].where(df_limpo['origem_agrupada'].isin(top_categorias), 'Outros')


# --- ETAPA 4: Gerar o novo gráfico limpo e focado ---
print("Analisando a Origem das Famílias (Agrupado e Focado nos Top 15)...")
plt.figure(figsize=(12, 8))
sns.countplot(
    y=df_limpo['origem_plot'],
    order=df_limpo['origem_plot'].value_counts().index # Ordena pela frequência
)
plt.title(f'Top {top_n} Origens das Famílias Entrevistadas')
plt.xlabel('Quantidade de Famílias')
plt.ylabel('Cidade/Estado de Origem')
plt.show()
```

# Análise 6 - Origem dos Rendimentos 💸🏠 e Número de Residentes 👨‍👩‍👧‍👦  
## 📊 Atende ao requisito: "IA para a origem dos rendimentos... verificando o número de residentes (média também)"

```python
# --- ETAPA 1: Definir as categorias de renda e suas palavras-chave ---
# Criamos um dicionário onde a chave é a nossa categoria limpa e os valores são as palavras que procuraremos.
categorias_renda = {
    'Trabalho Formal': 'carteira assinada|servidor|funcionário|formal',
    'Trabalho Informal': 'informal|bico|diarista',
    'Aposentadoria/Pensão': 'aposentadoria|aposentado|pensão|pensionista',
    'Autônomo/Empresário': 'autônomo|autônoma|empresa|empresário|dono de|micro',
    'Benefício Social': 'benefício social|bolsa família|auxílio',
    'Outras Fontes': 'aluguel|locação|ovos|venda'
}

# --- ETAPA 2: Criar uma coluna para cada categoria, marcando 1 se a família tem aquela renda ---
# Para cada categoria, verificamos se o texto na coluna original contém alguma das palavras-chave.
# O 'case=False' ignora se está em maiúsculas ou minúsculas.
# O '.astype(int)' transforma o resultado (True/False) em 1 ou 0.
for categoria, palavras_chave in categorias_renda.items():
    df_limpo[categoria] = df_limpo['origem_rendimentos'].str.contains(palavras_chave, case=False, na=False).astype(int)

# --- ETAPA 3: Calcular os totais para cada categoria ---
# Somamos os '1's de cada coluna para saber o total de famílias em cada categoria de renda.
contagem_renda = df_limpo[categorias_renda.keys()].sum().sort_values(ascending=False)


# --- ETAPA 4: Gerar o novo gráfico com os totais de cada fonte de renda ---
print("Analisando a Origem dos Rendimentos (por componentes)...")
plt.figure(figsize=(12, 8))

# Usamos o barplot, que é ideal para plotar valores já calculados (como nossa contagem).
sns.barplot(x=contagem_renda.values, y=contagem_renda.index)

plt.title('Componentes da Renda Familiar')
plt.xlabel('Número de Famílias')
plt.ylabel('Fonte de Renda Principal')
plt.show()


# A análise de número de residentes pode continuar a mesma
print("\n--- Análise do Número de Residentes por Domicílio ---")
print(df_limpo['num_residentes'].describe())
```

# Análise 10 - Refeições Diárias 🍽️  
## Quantidade de refeições feitas por dia pela família 👨‍👩‍👧‍👦

```python
# --- ETAPA 1: Importar a biblioteca de expressões regulares ---
import re

# --- ETAPA 2: Criar uma função para agrupar as respostas de refeições ---
def agrupar_refeicoes(texto):
    if not isinstance(texto, str):
        # ALTERAÇÃO 1: Unificar a categoria de não informados
        return 'Não informado/Outros'

    texto_lower = texto.lower()

    numeros = re.findall(r'\d+', texto_lower)
    if numeros:
        # ALTERAÇÃO 2: Converter para int e depois para str para remover zeros à esquerda (ex: '03' -> '3')
        return str(int(numeros[0]))

    if 'quatro' in texto_lower: return '4'
    if 'três' in texto_lower or ('café' in texto_lower and 'almoço' in texto_lower and 'janta' in texto_lower): return '3'
    if 'duas' in texto_lower or 'dois' in texto_lower: return '2'
    if 'uma' in texto_lower or 'um' in texto_lower: return '1'
    if 'cinco' in texto_lower: return '5'

    # ALTERAÇÃO 3: Unificar a categoria de não informados
    return 'Não informado/Outros'

# --- ETAPA 3: Aplicar a função para criar a nova coluna ---
df_limpo['refeicoes_agrupadas'] = df_limpo['refeicoes_por_dia'].apply(agrupar_refeicoes)

# --- ETAPA 4: Gerar o gráfico com ordenação numérica correta ---
print("Analisando o Número de Refeições Diárias (Agrupado e Ordenado)...")

# Lógica de ordenação aprimorada para garantir a ordem numérica
categorias = df_limpo['refeicoes_agrupadas'].unique()
numericas = sorted([int(c) for c in categorias if c.isdigit()])
textuais = sorted([c for c in categorias if not c.isdigit()])
ordem_final = [str(n) for n in numericas] + textuais # Junta as listas, com os números ordenados primeiro

plt.figure(figsize=(10, 6))
sns.countplot(
    x=df_limpo['refeicoes_agrupadas'],
    order=ordem_final # Usa a nova ordem final
)
plt.title('Número de Refeições Diárias por Família')
plt.xlabel('Refeições por Dia')
plt.ylabel('Quantidade de Famílias')
plt.show()

# Contagem em formato de tabela
print("\n--- Contagem de Refeições Diárias Agrupadas ---")
print(df_limpo['refeicoes_agrupadas'].value_counts())
```

# Análise 11 - Tempo de Moradia 🏠⏳ e Local Anterior 📍  
## Gráfico: "Há quanto tempo mora aqui?"  
## IA: "Onde morava antes?"

```python
# --- ETAPA 1: Importar a biblioteca de expressões regulares (para extrair números) ---
import re

# --- ETAPA 2: Criar uma função para agrupar as respostas ---
def agrupar_tempo_moradia(texto):
    # Se o valor for nulo ou não for texto, retorna uma categoria padrão
    if not isinstance(texto, str):
        return 'Não informado'

    texto_lower = texto.lower() # Converte para minúsculas para facilitar a busca

    # Verifica se a pessoa mora desde que nasceu
    if 'nasc' in texto_lower or 'vida toda' in texto_lower:
        return 'Desde que nasceu'

    # Tenta encontrar qualquer número no texto (ex: "moro há 25 anos")
    numeros = re.findall(r'\d+', texto)
    if numeros:
        anos = int(numeros[0]) # Pega o primeiro número encontrado
        if anos <= 5:
            return '0-5 anos'
        elif anos <= 10:
            return '6-10 anos'
        elif anos <= 20:
            return '11-20 anos'
        else:
            return 'Mais de 20 anos'

    # Se não encontrar um padrão conhecido, retorna 'Outros'
    return 'Outros/Não especificado'

# --- ETAPA 3: Aplicar a função para criar uma nova coluna com os dados limpos ---
df_limpo['tempo_moradia_agrupado'] = df_limpo['tempo_mora_aqui'].apply(agrupar_tempo_moradia)


# --- ETAPA 4: Gerar o novo gráfico usando a coluna agrupada ---
print("Analisando o Tempo de Moradia no Bairro (Agrupado)...")

# Define uma ordem lógica para as barras do gráfico
ordem_grafico = [
    '0-5 anos',
    '6-10 anos',
    '11-20 anos',
    'Mais de 20 anos',
    'Desde que nasceu',
    'Outros/Não especificado',
    'Não informado'
]

plt.figure(figsize=(12, 7))
sns.countplot(
    y=df_limpo['tempo_moradia_agrupado'],
    order=ordem_grafico # Usa a ordem que definimos
)
plt.title('Tempo de Moradia no Bairro (Agrupado)')
plt.xlabel('Quantidade de Famílias', fontsize=12)
plt.ylabel('Tempo de Moradia', fontsize=12)
plt.show()

# A análise de "Onde morava antes?" continua a mesma
print("\n--- Onde os moradores viviam antes? (Top 10) ---")
print(df_limpo['morava_antes'].value_counts().head(10))
```
